# GridSearch

Parámetros a optimizar:
- num_topics: número de tópicos (clusters)
- chunksize: número de documentos q se usarán en cada fragmento de training
- passes: número de pases a través del corpus durante el training
- alpha
- eta

```python
from gensim.models import CoherenceModel
from sklearn.metrics import davies_bouldin_score
from itertools import product
from scipy.spatial.distance import jaccard
```

```python
# espacio para buscar 

parametros = {
    'num_topics': range(2,38),
    'alpha': ['symmetric','asymmetric','auto'],
    'eta': ['symmetric','auto',None],
    'passes': [10, 20, 30],
    'chunksize': [100, 200]
}
```

```python
# guardo todo en un df

df_gridsearch = pd.DataFrame(columns = ['num_topics','alpha','eta','passes','chunksize','coherence','davies_bouldin'])
```

```python
def distancia_jaccard(modelo, num_topicos):
    topic_words = []

    # Recolectar palabras para cada tema
    for topic_id in range(num_topicos):
        try:
            palabras = [palabra for palabra, prob in modelo.show_topic(topic_id, topn=10)]
            topic_words.append(set(palabras))
        except IndexError:
            # Manejar el caso en que no haya suficientes temas generados
            topic_words.append(set())

    # Verificar que haya al menos dos temas válidos para calcular Jaccard
    if len(topic_words) < 2:
        return np.nan  # No es posible calcular Jaccard con menos de 2 temas

    # Calcular las distancias de Jaccard
    d_jaccard = []
    for i in range(len(topic_words)):
        for j in range(i + 1, len(topic_words)):
            d_jaccard.append(jaccard(topic_words[i], topic_words[j]))

    # Retornar el promedio de las distancias de Jaccard
    return np.mean(d_jaccard) if d_jaccard else np.nan
```

```python
from tqdm import tqdm
```

```python
combinaciones = list(product(parametros['num_topics'], parametros['alpha'], parametros['eta'], parametros['passes'], parametros['chunksize']))
```

```python
len(combinaciones)
```

```python
for p in tqdm(combinaciones, desc="GridSearch en progreso"):
    
    num_topics, alpha, eta, passes, chunksize = p
    
    print(f"Evaluando: num_topics={num_topics}, alpha={alpha}, eta={eta}, passes={passes}, chunksize={chunksize}")
    
    lda_model = LdaModel(corpus=doc_term_matrix, id2word=diccionario, random_state=2024,  # semilla
                         num_topics=num_topics,
                         alpha=alpha,
                         eta=eta,
                         passes=passes,
                         chunksize=chunksize)
    
    # métricas
    # 1. coherencia
    coherence_model = CoherenceModel(model=lda_model, texts=Documentos, dictionary=diccionario, coherence='c_v')
    coherence_score = coherence_model.get_coherence()
    
    # 2. davies bouldin
    doc_topic_dist = np.array([[prob for _, prob in lda_model.get_document_topics(doc, minimum_probability=0)] for doc in doc_term_matrix])
    db_score = davies_bouldin_score(doc_topic_dist, np.argmax(doc_topic_dist, axis=1))
    
    # 3. indice de jaccard
    #jaccard_score = distancia_jaccard(lda_model, num_topics)
    
    new_row = pd.DataFrame([{
        'num_topics': num_topics,
        'alpha': alpha,
        'eta': eta,
        'passes': passes,
        'chunksize': chunksize,
        'coherence': coherence_score,
        'davies_bouldin': db_score
    }])
    
    # Concatenar al DataFrame principal
    df_gridsearch = pd.concat([df_gridsearch, new_row], ignore_index=True)
```
